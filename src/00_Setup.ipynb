### **BóSight: Data Setup**  
#### Downloads MmCows from HuggingFace and extracts it to Google Drive.  
#### Input: none (fetches from HuggingFace Hub).  
#### Output: extracted dataset under /content/drive/MyDrive/MmCows/extracted/
#### **Note:** paths below assume a Colab environment with Google Drive mounted
#### at /content/drive. Update BASE if running elsewhere.


```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Define paths and the files to download
import os

BASE = "/content/drive/MyDrive/MmCows"
EXTRACT_DIR = f"{BASE}/extracted"
DOWNLOAD_DIR = "/content/dl"
os.makedirs(EXTRACT_DIR, exist_ok=True)
os.makedirs(DOWNLOAD_DIR, exist_ok=True)

# Each entry: (filename in HuggingFace repo, expected top-level folder name after extraction)
FILES = [
    ("visual_data.zip", "visual_data"),
    ("sensor_data.zip", "sensor_data"),
    ("cow_gallery.zip", "cows_gallery"),  # zip file is named "cow", extracted folder is "cows"
]
```

```python
# Install the HuggingFace Hub client
!pip install -q huggingface_hub
from huggingface_hub import hf_hub_download
```

```python
# Download and extract each file. Skip any folder that is already extracted.
import zipfile

for zip_name, folder_name in FILES:
    target = f"{EXTRACT_DIR}/{folder_name}"
    if os.path.isdir(target) and len(os.listdir(target)) > 0:
        print(f"{folder_name}/ already extracted, skipping.")
        continue

    print(f"Downloading {zip_name}...")
    zip_path = hf_hub_download(
        repo_id="neis-lab/mmcows",
        filename=zip_name,
        repo_type="dataset",
        local_dir=DOWNLOAD_DIR,
    )

    print(f"Extracting {zip_name} to Drive (this may take a while for visual_data)...")
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(EXTRACT_DIR)

    os.remove(zip_path)
    print(f"{folder_name}/ done.")
```
